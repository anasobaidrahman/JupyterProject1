```python
from random import Random
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
import kagglehub as kagglehub
import pandas as pd
import sklearn as sklearn
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from kagglehub import KaggleDatasetAdapter

from charset_normalizer import detect
import pandas as pd
import os
# Set the path to the file you'd like to load
file_path = "synthetic_dataset.csv"

# Load the latest version
df = kagglehub.dataset_load(
  KaggleDatasetAdapter.PANDAS,
  "himelsarder/retail-product-dataset-with-missing-values",
  file_path,
  # Provide any additional arguments like
  # sql_query or pandas_kwargs. See the
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

# If load_dataset returned a DataFrame, use it directly.
if isinstance(df, pd.DataFrame):
    product_data = df
else:
    # Otherwise try to read the file from disk. Check existence first and report helpful error.
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"File '{file_path}' not found. Make sure kagglehub saved the file to the current working directory ('{os.getcwd()}') or update file_path to the correct location.")

    with open(file_path, "rb") as f:
        raw_data = f.read()
        result = detect(raw_data)

    print(result)

    product_data = pd.read_csv(
        file_path,
        encoding=result["encoding"] or "latin1",   # fall back if detection is unsure
        encoding_errors="replace"                   # never crash; swap bad bytes with '' instead
    )

product_data
```

## Missing Value Cleaning

```python
product_data.isna().sum()
```

```python

product_data.dtypes
```

```python
# defining stretegies for transforming the missing values
stock_fil = SimpleImputer(strategy='constant',fill_value='in stock')
Category_fil = SimpleImputer(strategy='most_frequent')
price_fil = SimpleImputer(strategy='mean')
rating_fil = SimpleImputer(strategy='median')
discount_fil = SimpleImputer(strategy='mean')
```

```python
# defining the column transformer to apply the transformations
# Add remainder='passthrough' to keep columns not specified in the transformer
imputer = ColumnTransformer([('two',Category_fil,['Category']),
                             ('three',price_fil,['Price']),
                             ('five',rating_fil,['Rating']),
                             ('four',discount_fil,['Discount']),
                             ('one',stock_fil,['Stock'])],
                             remainder='passthrough')  # Keep other  columns unchanged and always give column names in list
transformed_data = imputer.fit_transform(product_data)

# If sparse, convert to dense array
# If the transformer produced a sparse matrix (common with encoders), convert to a dense NumPy array.
# `hasattr(transformed_data, 'toarray')` checks whether the result object provides a `toarray` method.
# Many sparse matrix types (e.g., scipy.sparse) implement `toarray()` to return a dense ndarray.
# hasattr method is used to check wheather the object has the attribute or not. If it has then it will convert the sparse matrix to dense array.
if hasattr(transformed_data, 'toarray'):
    transformed_data = transformed_data.toarray()

# Create DataFrame with proper column handling
data_filled = pd.DataFrame(transformed_data, columns=product_data.columns)
data_filled
```
