```python
from random import Random

import kagglehub as kagglehub
import pandas as pd
import sklearn as sklearn
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from kagglehub import KaggleDatasetAdapter

from charset_normalizer import detect
import pandas as pd

with open(csv_path, "rb") as f:
    raw_data = f.read()
    result = detect(raw_data)

print(result)

mobile_data = pd.read_csv(
    csv_path,
    encoding=result["encoding"] or "latin1",   # fall back if detection is unsure
    encoding_errors="replace"                   # never crash; swap bad bytes with '�' instead
)
mobile_data.head()
```

```python
mobile_data.drop(columns=['Launched Price (Dubai)','Launched Price (USA)','Launched Price (China)','Launched Price (India)'], inplace=True, errors='ignore')
mobile_data['Launched Price (Pakistan)'] = mobile_data['Launched Price (Pakistan)'].str.replace('PKR','').str.replace(',','').astype(str)
mobile_data
```

```python
x = mobile_data.drop(columns=['Launched Price (Pakistan)'])
y = mobile_data['Launched Price (Pakistan)']
# x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2,)
```

```python
# To transform string object into numeric by encoding and the transforming
from sklearn.preprocessing import OneHotEncoder #onehotencoder is used to encode
from sklearn.compose import ColumnTransformer #transformer is used to transform the data
feater_cat = x.select_dtypes(include=['object']).columns.tolist() #by using stect we selected string objects of columns and then it is converted to list by tolist
hottie = OneHotEncoder() # hottie is used to encode the string objects into numeric values
transformer = ColumnTransformer([('encoder', hottie, feater_cat)], remainder='passthrough') #transfromer is created that is created from column transformer and give argument encoder(the name of transformer) and the hottie(transformer itself) and the list of featers and the second argument is saying that the other featerus you would passthrough.
transformed_x = transformer.fit_transform(x).toarray() # at that the transformed x data is created by using the transformer and fit_transform method is used to fit the data and transform it into numeric values and then it is converted into array by using toarray method.
df = pd.DataFrame(transformed_x) # the transformed data is converted into data fram to visualize.
df
```

```python
# 2nd method to transform string object into numeric by dummy library of pandas
transformed_new = pd.get_dummies(mobile_data, columns=mobile_data.select_dtypes(include=['object']).columns.tolist()).astype(int) # the get_dummies method is used to transform the string object into numeric values and .astype(int) converts boolean True/False to numeric 1/0
transformed_new
```

```python
# lets split and fit the data to model
x_train, x_test, y_train, y_test = train_test_split(transformed_new, y, test_size=0.2, random_state=42) #data is splitted into train and test data by using train_test_split method and the test_size is 0.2 means 20% of data is used for testing and random_state is used to get the same result every time.
mobile_model = RandomForestClassifier(n_estimators=100, random_state=42) # the model is created by using RandomForestClassifier and n_estimators is the number of trees in the forest and random_state is used to get the same result every time.
mobile_model.fit(x_train, y_train) # the model is fitted to the train data by using fit method and the train data is passed as argument.
mobile_model.score(x_test, y_test) # the model is scored by using score method and the test data is passed as argument.
```
