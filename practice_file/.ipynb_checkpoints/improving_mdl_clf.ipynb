```python
from holoviews.element import comparison
from sklearn.ensemble import RandomForestClassifier as clf
import pandas as pd
from sklearn.model_selection import train_test_split
import pickle
```

```python
data = pd.read_csv('C:\\Users\\HP\\PycharmProjects\\JupyterProject1\\practice_file\\heart.csv')
data_X = data.drop(columns=['target'])
data_y = data['target']
```

# Improving the Model by Hyperparameter Tuning

```python
# Split the data into training, validation, and test sets (70% train, 15% validation, 15% test)
trainX, tempX, trainY, tempY = train_test_split(data_X, data_y, stratify=data_y, test_size=0.3, random_state=42)# split data into train and temporary set and stratify preserves the portions of target value in every split.
# Split the temporary set into validation and test sets (15% each)
valX, testX, valY, testY = train_test_split(tempX, tempY, stratify=tempY, test_size=0.5, random_state=42)
len(trainX), len(valX), len(testX)  # Check the lengths of the splits
```

```python
# Create a Random Forest Classifier with hyperparameter tuning
my_mdl = clf(random_state=42) # Create a Random Forest Classifier with default parameters
# Fit the model on the training data
my_mdl.fit(trainX, trainY) # Fit the model on the training data
# Evaluate the model on the validation set
val_score = my_mdl.score(valX, valY) # Evaluate the model on the validation set
print("Validation Score:", val_score) # Print the validation score
```

```python
def evaluate_model(model, trainX, trainY, valX, valY, testX, testY):
    """
    Evaluate the model on training, validation, and test sets.
    Returns a dictionary with scores for each set.
    """
    scores = {
        'train_score': model.score(trainX, trainY),
        'val_score': model.score(valX, valY),
        'test_score': model.score(testX, testY)
    }
    return scores
def evaluate_model_hyperparameter(model,testX,testY):
    """
    Evaluate the model using different evaluation metrics as sklearn functions.
    """
    from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
    y_pred =model.predict(testX)
    print("Accuracy:",round(accuracy_score(testY, y_pred), 2))
    print("Precision:", round(precision_score(testY, y_pred), 2))
    print("Recall:", round(recall_score(testY, y_pred), 2))
    print("F1 Score:", round(f1_score(testY, y_pred), 2))
    print("ROC AUC Score:", round(roc_auc_score(testY, y_pred), 2))

    results_score = {
        'accuracy': accuracy_score(testY, y_pred),
        'precision': precision_score(testY, y_pred),
        'recall': recall_score(testY, y_pred),
        'f1_score': f1_score(testY, y_pred),
        'roc_auc_score': roc_auc_score(testY, y_pred)
    }
    return results_score
```

```python
baseline_results=evaluate_model_hyperparameter(my_mdl,valX,valY);
```

#### Setting hyperparameters Manually

```python
# Create a Random Forest Classifier with specific hyperparameters
my_mdl_tuned = clf(n_estimators=5, min_samples_split=4,min_samples_leaf=4,max_features= 'sqrt',max_depth=None,bootstrap=False,random_state = 42)
 # Create a Random Forest Classifier with specific hyperparameters
# Fit the model on the training data
my_mdl_tuned.fit(trainX, trainY) # Fit the model on the training data
# Evaluate the model on the validation set
manul_trn_result=evaluate_model_hyperparameter(my_mdl_tuned,valX,valY) # Evaluate the model on the validation set
```

### Tuning hyperparameters using RandomizedSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
# Define the hyperparameter grid for RandomizedSearchCV
param_grid = {
    'n_estimators': [50, 100, 200, 300, 400, 500],
    'max_depth': [None, 5, 10, 15, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False],
    'max_features': ['sqrt', 'log2', None],
}
# Create a Random Forest Classifier
rf_model = clf(random_state=42)
# Create a RandomizedSearchCV object
random_search = RandomizedSearchCV(estimator=rf_model, param_distributions=param_grid, n_iter=20,n_jobs=-1, cv=5,verbose=2, random_state=42,)
# Fit the RandomizedSearchCV object on the training data
random_search.fit(trainX, trainY)
```

```python
random_search.score(valX, valY)  # Evaluate the best model on the validation set
```

```python
random_search_resul=evaluate_model_hyperparameter(random_search.best_estimator_,valX,valY)  # Evaluate the best model on the validation set using different metrics
```

```python
random_search.best_params_
```

### Tuning hyperparameters using GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
# Define the hyperparameter grid for GridSearchCV
param_grid = {
    'n_estimators': [10,100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 4],
    'min_samples_leaf': [1, 2],
    'bootstrap': [True, False],
    'max_features': ['sqrt', 'log2', None],
    'warm_start': [True]
}
# Create a Random Forest Classifier
rf_model = clf(random_state=42)
# Create a GridSearchCV object
grid_search = GridSearchCV(estimator=rf_model, param_grid=param_grid, cv=5, n_jobs=-1, verbose=2) # Configure the GridSearchCV object with the model, hyperparameter grid, cross-validation folds, and verbosity
# Fit the GridSearchCV object on the training data
grid_search.fit(trainX, trainY) # The actual fitting of the model on the training data using the specified hyperparameter grid and cross-validation
```

```python
grid_search.score(valX, valY)  # Evaluate the best model on the validation set
```

```python
grid_search_result=evaluate_model_hyperparameter(grid_search.best_estimator_,valX,valY)  # Evaluate the best model on the validation set using different metrics
```

```python
grid_search.best_params_
```

### Ploting all the hyperparameter tuning results

```python
import matplotlib.pyplot as plt
import numpy as np
# Get the results from the GridSearchCV
results = pd.DataFrame({'Base_line': baseline_results, 'Manual':manul_trn_result, 'RandomizedSearchCV': random_search_resul, 'GridSearchCV': grid_search_result})
results
```

```python
results.plot(kind='bar', figsize=(12, 6))
```

### Saving the best model by using pickle


```python
# Save the best model from GridSearchCV to a file using pickle
import pickle
pickle.dump(grid_search, open('grid_search.pkl', 'wb'))
```

```python
# Load the saved model from the file using pickle
loaded_model = pickle.load(open('grid_search.pkl', 'rb'))
# Evaluate the loaded model on the test set
evaluate_model_hyperparameter(loaded_model, testX, testY);
```

```python
import joblib
# Save the best model from GridSearchCV to a file using joblib
joblib.dump(grid_search, 'grid_search_joblib.joblib')
```

```python
# Load the saved model from the file using joblib
loaded_model_joblib = joblib.load('grid_search_joblib.joblib')
# Evaluate the loaded model on the test set
evaluate_model_hyperparameter(loaded_model_joblib, testX, testY);
```
